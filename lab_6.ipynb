```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

# Create dataset
data = {
    'Study_Hours': [2, 4, 6, 8, 10],
    'Attendance': [50, 60, 70, 80, 90],
    'Marks': [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)

# 1. Display Original Data
print("1. Original Data:")
print(df)

# 2. Data Normalization using MinMaxScaler
normalizer = MinMaxScaler()
normalized_data = normalizer.fit_transform(df)

normalized_df = pd.DataFrame(
    normalized_data,
    columns=df.columns
)

print("\n2. Normalized Data:")
print(normalized_df.round(2))

# 3. Data Standardization using StandardScaler
standardizer = StandardScaler()
standardized_data = standardizer.fit_transform(df)

standardized_df = pd.DataFrame(
    standardized_data,
    columns=df.columns
)

print("\n3. Standardized Data:")
print(standardized_df.round(2))

# 4. Principal Component Analysis (PCA)
pca = PCA(n_components=2)
pca_data = pca.fit_transform(standardized_data)

pca_df = pd.DataFrame(
    pca_data,
    columns=['PC1', 'PC2']
)

print("\n4. Data after PCA:")
print(pca_df.round(2))

# 5. Display Explained Variance
print("\n5. Explained Variance Ratio:")
print(pca.explained_variance_ratio_.round(4))
```
