```python
import pandas as pd
data = {
    "Student": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 9, 4, 5],
    "Attendance": [65, 70, 72, 75, 78, 80, 82, 85, 68, 74],
    "Assignment_Marks": [12, 14, 15, 17, 18, 19, 20, 22, 14, 16],
    "Internal_Marks": [18, 20, 22, 24, 25, 27, 28, 30, 21, 31],
    "Final_Marks": [45, 50, 55, 60, 65, 70, 75, 80, 52, 58]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display dataset
df
```

```python
import matplotlib.pyplot as plt
plt.hist(df["Final_Marks"], bins=5)
plt.xlabel("Final Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Final Marks")
plt.show()
```

```python
plt.boxplot(df["Final_Marks"])
plt.xlabel("Final Marks")
plt.ylabel("Marks")
plt.title("Box Plot of Final Marks")
plt.show()
```

```python
plt.boxplot([
    df["Study_Hours"],
    df["Attendance"],
    df["Assignment_Marks"],
    df["Internal_Marks"],
    df["Final_Marks"]
])
plt.xticks(
    [1, 2, 3, 4, 5],
    ["Study Hours", "Attendance", "Assignment", "Internal", "Final"]
)
plt.ylabel("Values")
plt.title("Box Plot of Student Performance")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.pairplot(
    df[
        [
            "Study_Hours",
            "Attendance",
            "Assignment_Marks",
            "Internal_Marks",
            "Final_Marks"
        ]
    ]
)
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
