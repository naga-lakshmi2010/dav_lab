```python
import numpy as np
import matplotlib.pyplot as plt

# Sample data
np.random.seed(42)
X = np.random.rand(50) * 10          # feature
y = 3 * X + 5 + np.random.randn(50) * 2   # target: y = 3x + 5 + noise

# Calculate slope (m) and intercept (b) using least squares
x_mean, y_mean = X.mean(), y.mean()
m = np.sum((X - x_mean) * (y - y_mean)) / np.sum((X - x_mean) ** 2)
b = y_mean - m * x_mean

print(f"Slope (m): {m:.4f}")
print(f"Intercept (b): {b:.4f}")

# Predictions
y_pred = m * X + b

# R² score
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - y_mean) ** 2)
r2 = 1 - ss_res / ss_tot
print(f"R² score: {r2:.4f}")

# Plot
plt.scatter(X, y, label="Data")
plt.plot(X, y_pred, color="red", label="Regression line")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Create dataset
data = pd.DataFrame({
    'Hours': [1, 2, 3, 4, 5, 6, 7, 8],
    'Marks': [35, 40, 50, 55, 65, 70, 75, 85]
})

# Build simple linear regression model
X = data[['Hours']]
y = data['Marks']

model = LinearRegression()
model.fit(X, y)

# Display model coefficients
print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])

# Predict values
pred = model.predict(X)
print("Predicted Marks:", pred)

# Calculate R-squared and RMSE
r2 = r2_score(y, pred)
rmse = np.sqrt(mean_squared_error(y, pred))

print("R-squared:", r2)
print("RMSE:", rmse)
# Plot regression line
plt.scatter(data['Hours'], y, color='blue', label='Actual Marks')
plt.plot(data['Hours'], pred, color='red', label='Regression Line')
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Simple Linear Regression")
plt.legend()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Create dataset
data = pd.DataFrame({
    'Hours': [1, 2, 3, 4, 5, 6, 7, 8],
    'Sleep': [7, 6, 8, 6, 7, 8, 7, 6],
    'Attendance': [80, 85, 75, 90, 85, 95, 90, 100],
    'Marks': [35, 40, 50, 55, 65, 70, 75, 85]
})

# Define independent and dependent variables
X = data[['Hours', 'Sleep', 'Attendance']]
y = data['Marks']

# Build multiple linear regression model
model = LinearRegression()
model.fit(X, y)

# Display model coefficients
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)

# Predict values
pred = model.predict(X)
print("Predicted Marks:", pred)
# Calculate R-squared and RMSE
r2 = r2_score(y, pred)
rmse = np.sqrt(mean_squared_error(y, pred))

print("R-squared:", r2)
print("RMSE:", rmse)

# Plot actual vs predicted marks
plt.scatter(y, pred, color='blue', label='Predicted Marks')
plt.plot([y.min(), y.max()],
         [y.min(), y.max()],
         color='red', label='Perfect Prediction')

plt.xlabel("Actual Marks")
plt.ylabel("Predicted Marks")
plt.title("Multiple Linear Regression")
plt.legend()
plt.show()
```
