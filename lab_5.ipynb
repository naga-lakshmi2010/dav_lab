```python
import pandas as pd

# Sample dataset
marks = [10, 12, 14, 16, 18, 20, 22, 24, 26, 100]

df = pd.DataFrame({"Marks": marks})

# Calculate quartiles
Q1 = df["Marks"].quantile(0.25)
Q2 = df["Marks"].quantile(0.50)
Q3 = df["Marks"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Print results
print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
import pandas as pd

# Sample data
values = [
    48, 49, 49, 50, 50, 50, 51, 51, 51, 52,
    52, 52, 53, 53, 53, 54, 54, 55, 55, 100
]

# Calculate mean and standard deviation
mean = sum(values) / len(values)
std = pd.Series(values).std(ddof=0)

print("Mean:", round(mean, 2))
print("Standard Deviation:", round(std, 2))

# Detect outliers
for x in values:
    z = (x - mean) / std
    if abs(z) > 3:
        print("Outlier:", x)
        print("Z-score:", round(z, 2))
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

# Create a dataset
data = {
    'Study_Hours': [2, 4, 6, 8, 10],
    'Attendance': [50, 60, 70, 80, 90],
    'Marks': [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)

print("1. Original Data:")
print(df)

# 2. NORMALIZATION
scaler = MinMaxScaler()
normalized_data = scaler.fit_transform(df)

normalized_df = pd.DataFrame(
    normalized_data,
    columns=df.columns
)

print("\n2. Normalized Data:")
print(normalized_df.round(2))

# 3. STANDARDIZATION
standardizer = StandardScaler()
standardized_data = standardizer.fit_transform(df)

standardized_df = pd.DataFrame(
    standardized_data,
    columns=df.columns
)

print("\n3. Standardized Data:")
print(standardized_df.round(2))

# 4. PCA DIMENSIONALITY REDUCTION
pca = PCA(n_components=2)
pca_data = pca.fit_transform(standardized_data)

pca_df = pd.DataFrame(
    pca_data,
    columns=['PC1', 'PC2']
)

print("\n4. PCA Dimensionality Reduction:")
print(pca_df.round(2))
```
